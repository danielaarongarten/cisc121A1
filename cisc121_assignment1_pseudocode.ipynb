## Question 1 — Cyclic Rotation

```text
BEGIN
    REPEAT K times
        IF array A is empty THEN
            RETURN A
        END IF
        Remove last element from A
        Insert removed element at beginning of A
    END REPEAT
    RETURN A
END
```

## Question 2 — Odd Occurrences in Array

```text
BEGIN
    Create an empty dictionary called counts
    FOR each number in array A
        IF number is already in counts THEN
            Add 1 to its count
        ELSE
            Set its count to 1
        END IF
    END FOR
    FOR each number in counts
        IF its count is odd THEN
            RETURN that number
        END IF
    END FOR
END
```

## Question 3 — Frog Jump

```text
BEGIN
    SET distance to Y - X
    SET jumps to distance // D
    IF distance is not divisible by D THEN
        Add 1 to jumps
    END IF
    RETURN jumps
END
```

## Question 4 — Missing Element

```text
BEGIN
    SET N to length of array + 1
    SET total to N * (N + 1) // 2
    SET actual to sum of array
    RETURN total - actual
END
```

## Question 6 — Permutation Check

```text
BEGIN
    SET N to length of array
    IF number of unique elements is not N THEN
        RETURN 0
    END IF
    IF smallest number in array is not 1 THEN
        RETURN 0
    END IF
    IF largest number in array is not N THEN
        RETURN 0
    END IF
    RETURN 1
END
```

CyclicRotation

```python
#define a function
def solution(A, K):
  #repeat the function k times
  for i in range(K):
    #check if aaray is empty, if it is return as is
    if len(A) == 0:
      return A
    #remove the last element from the list
    last = A.pop()
    #insert last element at the very beginning
    A.insert(0, last)
  #return the new array
  return A
```

```python

def solution(A):
    #Create an empty dictionary
    counts = {}

    #Go through each number in A
    for num in A:

        #If the number is already in the dictionary, add 1
        if num in counts:
            counts[num] += 1

        #Otherwise add the number to the dictionary
        else:
            counts[num] = 1

    #Go through the numbers in the dictionary
    for num in counts:

        #Check if the count is odd
        if counts[num] % 2 != 0:

            #Return the number with an odd count
            return num

```

```python
solution([9,3,9,3,9,7,9])
```

question 3

```python
def solution(X,Y,D):
  #find distance needed and number jumps
  distance = Y - X
  jumps = distance // D

  #add one more jump if not == to 0
  if distance % D != 0:
    jumps += 1
  return jumps
```

```python
solution(10,100,5)
```

```python
def solution(A):
  #find how many numbers should in array
  N = len(A) + 1
  #find the total sum
  total = N * (N+1)//2
  #find actual sum
  actual = sum(A)

  return total - actual
```

```python
solution([1,2,3])
```

```python
def solution(A):
  #find length of the array
  N = len(A)
  #check if their any duplicate numbers
  if len(set(A)) != N:
    return 0
  #check if smallest number is 1
  if min(A) != 1:
    return 0
  #check if biggest number is N
  if max(A) != N:
    return 0
  #return 1 if all is good
  return 1
```

```python
solution([4,1,3,2])
```
